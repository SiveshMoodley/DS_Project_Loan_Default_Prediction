## **Data Description:**
The Home Equity dataset (HMEQ) contains baseline and loan performance information for 5,960 recent home equity loans. The target (BAD) is a binary variable that indicates whether an applicant has ultimately defaulted or has been severely delinquent. This adverse outcome occurred in 1,189 cases (20 percent). 12 input variables were registered for each applicant.


* **BAD:** 1 = Client defaulted on loan, 0 = loan repaid

* **LOAN:** Amount of loan approved.

* **MORTDUE:** Amount due on the existing mortgage.

* **VALUE:** Current value of the property.

* **REASON:** Reason for the loan request. (HomeImp = home improvement, DebtCon= debt consolidation which means taking out a new loan to pay off other liabilities and consumer debts)

* **JOB:** The type of job that loan applicant has such as manager, self, etc.

* **YOJ:** Years at present job.

* **DEROG:** Number of major derogatory reports (which indicates a serious delinquency or late payments).

* **DELINQ:** Number of delinquent credit lines (a line of credit becomes delinquent when a borrower does not make the minimum required payments 30 to 60 days past the day on which the payments were due).

* **CLAGE:** Age of the oldest credit line in months.

* **NINQ:** Number of recent credit inquiries.

* **CLNO:** Number of existing credit lines.

* **DEBTINC:** Debt-to-income ratio (all your monthly debt payments divided by your gross monthly income. This number is one way lenders measure your ability to manage the monthly payments to repay the money you plan to borrow.

```python
# Define a function for univariate analysis with boxplot and histogram
def univariate_analysis(data, column, bins=30):
    """
    Perform univariate analysis on a numerical column, displaying both a boxplot and histogram.

    Parameters:
        data (pd.DataFrame): The dataset.
        column (str): The column name for analysis.
        bins (int): Number of bins for the histogram.
    """
    fig, axes = plt.subplots(2, 1, figsize=(10, 8), gridspec_kw={'height_ratios': [1, 4]})

    # Boxplot
    sns.boxplot(data=data, x=column, ax=axes[0], color='violet', fliersize=2)
    axes[0].set_title(f'Boxplot of {column}', fontsize=12)
    axes[0].set_xlabel('')

    # Histogram
    sns.histplot(data[column], kde=True, bins=bins, ax=axes[1], color='blue', alpha=0.7)
    axes[1].axvline(data[column].median(), color='green', linestyle='-', label=f"Median: {data[column].median():.2f}")
    axes[1].axvline(data[column].mean(), color='red', linestyle='--', label=f"Mean: {data[column].mean():.2f}")
    axes[1].set_title(f'Histogram of {column}', fontsize=12)
    axes[1].set_xlabel(column)
    axes[1].legend()

    plt.tight_layout()
    plt.show()

# Perform univariate analysis on multiple numerical columns
numerical_columns = ['LOAN', 'MORTDUE', 'VALUE', 'YOJ', 'DEROG', 'DELINQ', 'CLAGE', 'NINQ', 'CLNO', 'DEBTINC']

for col in numerical_columns:
    print(f"\nAnalyzing {col}:")
    univariate_analysis(data, col)
```

### **Bivariate Analysis**

```python
from scipy.stats import ttest_ind, chi2_contingency

# Define a function for bivariate analysis
def bivariate_analysis(data, target, column):
    """
    Perform bivariate analysis between a feature and the target variable.

    Parameters:
        data (pd.DataFrame): The dataset.
        target (str): The target variable (e.g., BAD).
        column (str): The column to analyze against the target.
    """
    print(f"\n--- Analyzing Relationship between {column} and {target} ---")

    if data[column].dtype == 'object' or data[column].nunique() < 10:  # Categorical variable
        # Calculate the proportions of default for each category
        proportions = data.groupby(column)[target].mean()
        print(f"Proportions of {target} by {column}:\n{proportions}")

        # Visualize with a bar plot
        plt.figure(figsize=(8, 4))
        sns.barplot(x=proportions.index, y=proportions.values, palette='viridis')
        plt.title(f'{target} Proportions by {column}')
        plt.xlabel(column)
        plt.ylabel(f'Proportion of {target}')
        plt.xticks(rotation=45)
        plt.show()

        # Perform a chi-square test for categorical variables
        contingency_table = pd.crosstab(data[column], data[target])
        chi2, p, _, _ = chi2_contingency(contingency_table)
        print(f"Chi-square test: Chi2 = {chi2:.4f}, p-value = {p:.4f}")

    else:  # Numerical variable
        # Split data into defaulted and non-defaulted groups
        group1 = data[data[target] == 1][column]  # Defaulted
        group2 = data[data[target] == 0][column]  # Non-defaulted

        # Summary statistics
        print(f"Defaulted {column} Mean: {group1.mean():.2f}, Std: {group1.std():.2f}")
        print(f"Non-defaulted {column} Mean: {group2.mean():.2f}, Std: {group2.std():.2f}")

        # Perform t-test
        t_stat, p_val = ttest_ind(group1, group2, nan_policy='omit')
        print(f"T-test: t-statistic = {t_stat:.4f}, p-value = {p_val:.4f}")

        # Visualize with boxplots and histograms
        plt.figure(figsize=(10, 5))
        plt.subplot(1, 2, 1)
        sns.boxplot(x=target, y=column, data=data, palette='muted')
        plt.title(f'Boxplot of {column} by {target}')

        plt.subplot(1, 2, 2)
        sns.histplot(group1, kde=True, color='red', label='Defaulted', alpha=0.6, bins=30)
        sns.histplot(group2, kde=True, color='blue', label='Non-defaulted', alpha=0.6, bins=30)
        plt.title(f'Distribution of {column} by {target}')
        plt.xlabel(column)
        plt.legend()
        plt.tight_layout()
        plt.show()

# Example: Apply bivariate analysis to relevant features
target_variable = 'BAD'
features = ['LOAN', 'MORTDUE', 'VALUE', 'REASON', 'JOB', 'YOJ', 'DEROG', 'DELINQ', 'CLAGE', 'NINQ', 'CLNO', 'DEBTINC']  # Replace with relevant columns

for feature in features:
    bivariate_analysis(data, target_variable, feature)
```

## Treating Outliers

```python
# Function to detect outliers using the IQR method
def detect_outliers_iqr(data, column):
    """
    Detects outliers in a column using the IQR method.

    Parameters:
        data (pd.DataFrame): The dataset.
        column (str): The column to check for outliers.

    Returns:
        Tuple of lower bound, upper bound, and a filtered DataFrame.
    """
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"{column}: Lower Bound = {lower_bound}, Upper Bound = {upper_bound}")
    return lower_bound, upper_bound

# Function to cap outliers
def cap_outliers(data, column):
    """
    Caps outliers in a column at the lower and upper bounds.

    Parameters:
        data (pd.DataFrame): The dataset.
        column (str): The column to cap outliers.

    Returns:
        Modified DataFrame with capped values.
    """
    lower_bound, upper_bound = detect_outliers_iqr(data, column)
    data[column] = np.where(data[column] < lower_bound, lower_bound, data[column])
    data[column] = np.where(data[column] > upper_bound, upper_bound, data[column])
    return data

# Function to process multiple columns
def treat_outliers(data, columns):
    """
    Detects and caps outliers for multiple columns in the dataset.

    Parameters:
        data (pd.DataFrame): The dataset.
        columns (list): List of columns to process.

    Returns:
        Modified DataFrame with capped outliers.
    """
    for column in columns:
        print(f"\nProcessing column: {column}")
        data = cap_outliers(data, column)
    return data

# Treat outliers for relevant numerical columns
numerical_columns = ['LOAN', 'MORTDUE', 'VALUE', 'YOJ', 'DEROG', 'DELINQ', 'CLAGE', 'NINQ', 'CLNO', 'DEBTINC']  # Replace with relevant columns
data = treat_outliers(data, numerical_columns)

# Verify the changes
print("\nOutlier treatment completed. Checking updated summary:")
print(data[numerical_columns].describe())
```

```python
# Import libraries for data preparation
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Data Preparation
# Separate features (X) and target variable (y)
X = data.drop(columns=['BAD'])  # Replace 'BAD' with the target variable in your dataset
y = data['BAD']

# Identify Numerical and Categorical Columns
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = X.select_dtypes(include=['object', 'category']).columns

# Define Preprocessing for Numerical Features
numerical_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())  # Standardizes numerical columns
])

# Define Preprocessing for Categorical Features
categorical_transformer = Pipeline(steps=[
    ('encoder', OneHotEncoder(handle_unknown='ignore'))  # Encodes categorical variables
])

# Combine Preprocessing in a ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ]
)

# Split Data into Training and Test Sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Apply Preprocessing
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

# Print Preprocessing Summary
print("Data Preprocessing Completed")
print(f"Numerical Columns: {numerical_cols.tolist()}")
print(f"Categorical Columns: {categorical_cols.tolist()}")
print(f"Processed Training Data Shape: {X_train_processed.shape}")
print(f"Processed Test Data Shape: {X_test_processed.shape}")
```

### **Decision Tree - Hyperparameter Tuning**

* Hyperparameter tuning is tricky in the sense that **there is no direct way to calculate how a change in the hyperparameter value will reduce the loss of your model**, so we usually resort to experimentation. We'll use Grid search to perform hyperparameter tuning.
* **Grid search is a tuning technique that attempts to compute the optimum values of hyperparameters.**
* **It is an exhaustive search** that is performed on the specific parameter values of a model.
* The parameters of the estimator/model used to apply these methods are **optimized by cross-validated grid-search** over a parameter grid.

**Criterion {“gini”, “entropy”}**

The function to measure the quality of a split. Supported criteria are “gini” for the Gini impurity and “entropy” for the information gain.

**max_depth**

The maximum depth of the tree. If None, then nodes are expanded until all leaves are pure or until all leaves contain less than min_samples_split samples.

**min_samples_leaf**

The minimum number of samples is required to be at a leaf node. A split point at any depth will only be considered if it leaves at least min_samples_leaf training samples in each of the left and right branches. This may have the effect of smoothing the model, especially in regression.

You can learn about more Hyperpapameters on this link and try to tune them.

https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html


```python
# Import libraries for hyperparamter tuning
from sklearn.model_selection import GridSearchCV

# Hyperparameter Tuning using GridSearchCV
param_grid = {
    'max_depth': [3, 5, 7, 10],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 5],
    'criterion': ['gini', 'entropy']
}

grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    scoring='roc_auc',
    cv=5,
    verbose=1,
    n_jobs=-1
)

# Fit Grid Search
grid_search.fit(X_train_processed, y_train)

# Best Parameters
print("Best Hyperparameters:", grid_search.best_params_)

# Evaluate the tuned model
best_model = grid_search.best_estimator_
y_pred_tuned = best_model.predict(X_test_processed)
y_pred_proba_tuned = best_model.predict_proba(X_test_processed)[:, 1]

print("\nTuned Decision Tree Model Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_tuned):.2f}")
print(f"ROC-AUC: {roc_auc_score(y_test, y_pred_proba_tuned):.2f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_tuned))
```

#### Observations from Hyperparameter Tuned Decision Tree Model

### **Building a Random Forest Classifier**

**Random Forest is a bagging algorithm where the base models are Decision Trees.** Samples are taken from the training data and on each sample a decision tree makes a prediction.

**The results from all the decision trees are combined together and the final prediction is made using voting or averaging.**

```python
# Import libraries for data preparation
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Data Preparation
# Separate features (X) and target variable (y)
X = data.drop(columns=['BAD'])  # Replace 'BAD' with the target variable in your dataset
y = data['BAD']

# Identify Numerical and Categorical Columns
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = X.select_dtypes(include=['object', 'category']).columns

# Define Preprocessing for Numerical Features
numerical_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())  # Standardizes numerical columns
])

# Define Preprocessing for Categorical Features
categorical_transformer = Pipeline(steps=[
    ('encoder', OneHotEncoder(handle_unknown='ignore'))  # Encodes categorical variables
])

# Combine Preprocessing in a ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ]
)

# Split Data into Training and Test Sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Apply Preprocessing
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

# Print Preprocessing Summary
print("Data Preprocessing Completed")
print(f"Numerical Columns: {numerical_cols.tolist()}")
print(f"Categorical Columns: {categorical_cols.tolist()}")
print(f"Processed Training Data Shape: {X_train_processed.shape}")
print(f"Processed Test Data Shape: {X_test_processed.shape}")
```
